# OH Analysis
## Preamble

```python
import warnings
from collections import Counter
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from chempiler import ChempilerTrajectory, plot_rdf, diffusive_window
from chempiler.state_engine import atom_hop, ligand_exchange

import tol_colors as tc
cset = tc.bright

plt.rcParams.update({
    'grid.alpha': 0.3,
    'font.size': 12,
    'axes.prop_cycle': plt.cycler(color=cset)
})

dt = 0.5 # fs
dump_interval = 10

LOGFILE     = '../data/MACE-POLAR-1-M/39_water_OH/neutral_doublet/39_water_OH.log'
REF_LOGFILE = '../data/MACE-POLAR-1-M/40_water/40_water_polar.log'
TRAJFILE    = '../data/MACE-POLAR-1-M/39_water_OH/neutral_doublet/39_water_OH.traj'
REF_TRAJFILE = '../data/MACE-POLAR-1-M/40_water/40_water_polar.traj'
```

```python
print('Reading OH• trajectory...')
traj = ChempilerTrajectory(
    TRAJFILE,
    mode='molecular', covalent_scale=0.8, # NOTE: The small covalent scale is important to capture the O-H bond in the ion. To avoid seeing clusters of water molecules as bonded.
)
print('Building topology...')
traj.build(
    cache_file='tests/OH_rad_dyn_cache.h5',
    persistence=3,
)
print('Done.')
print('Molecular species in trajectory:')
traj.summary()
```

```python
print('Reading water trajectory...')
traj_ref = ChempilerTrajectory(
    REF_TRAJFILE,
    mode='molecular',
    covalent_scale=0.8,
)
print('Building topology...')
traj_ref.build(
    cache_file='tests/40_water_cache.h5',
    persistence=3,
)
print('Molecular species in trajectory:')
traj_ref.summary()
```

## Summary of molecules

```python
# Dataframe
rows = []
for formula, _ in traj.summary().items():
    segs = traj.lifetime_segments(formula)
    lengths = np.array([e - s for s, e in segs])
    covered = lengths.sum()
    rows.append({
        'formula':  formula,
        'n_segs':   len(segs),
        'coverage': f"{100 * covered / len(traj.frames):.1f}%",
        'min':      int(lengths.min()),
        'median':   int(np.median(lengths)),
        'mean':     f"{lengths.mean():.1f}",
        'max':      int(lengths.max()),
    })

species = sorted(traj.summary().keys(), key=lambda f: -traj.summary()[f])
fig, ax = plt.subplots(figsize=(20, 0.5 * len(species) + 1))

for i, (row_colour, formula) in enumerate( zip(cset, species) ):
    for s, e in traj.lifetime_segments(formula):
        ax.barh(i, e - s, left=s, height=0.6, color=row_colour, alpha=0.8)

ax.set_yticks(range(len(species)))
ax.set_yticklabels(species)
ax.set_xlabel('Frame')
ax.set_xlim(0, len(traj.frames))
plt.tight_layout()

pd.DataFrame(rows).set_index('formula')
```

## RDFs

```python
from ase.io import write as ase_write
from chempiler import rdf_shells


plts = {
    'O$_h \\cdots$O$_w$': ({"HO": "O"}, {"H2O": "O"}),
    'O$_h \\cdots$H$_w$': ({"HO": "O"}, {"H2O": "H"}),
    'H$_h \\cdots$O$_w$': ({"HO": "H"}, {"H2O": "O"}),
}
fig, axs = plt.subplots(1,3, figsize=(3*5, 4))
for ax, (label, pair) in zip(axs, plts.items()):
    CTR, TGT = pair
    r, g, n = traj.rdf(center=CTR, target=TGT, dr=0.05, integrate=True)
    shells = rdf_shells(r, g, n, smooth_sigma=0.2)

    peaks = [
        str( peak_value )
        for peak_value in shells.peaks.round(decimals=3)
    
    ]; print('  Peaks:  ', ',  '.join(peaks))
    cns = [
        str( cn )
        for cn in shells.cns.round(decimals=3)
    ] ; print('  CNs:    ', ',  '.join(cns))

    ax.plot(r,g)
    ax.plot(r,n, ':')

    ax.set_title(label)
    ax.set_xlabel('r /Å')
    ax.set_ylim(0,20)
    ax.grid(axis='y')
axs[0].set_ylabel('$g(r) / n(r)$')
```

### H in HO to O in H2O
Since the insets are cumulative (inset 3 contains inset 2, inset 2 contains inset 1 and so on), this figure shows that the coordination on top of the second inset is more contracted.

```python
from chempiler import show
from scipy.ndimage import gaussian_filter1d


r, g = traj.rdf(center={"HO": "H"}, target={"H2O": "O"}, dr=0.05)
dr = float(r[1] - r[0])
g_smooth = gaussian_filter1d(g, sigma=0.2/dr)
g_curv = np.clip(-np.gradient(np.gradient(g_smooth, r), r), 0, None)
g_curv /= g_curv.max()

envs_shoulder = traj.rdf_shell_environments(
    center={"HO": "H"}, target={"H2O": "O"},
    rdf=(r, g_curv),
    n_per_shell=5, stride=5,smooth_sigma=0.0,
    peak_min_g=0.01,
)
print("All detected positions:", list(envs_shoulder.keys()))

fig, ax = plt.subplots()
plot_rdf(r, g_curv, ax=ax, insets={
        peak-0.5: ats[0] for peak, ats in envs_shoulder.items()
}, atomic_scale=1.0, inset_y=0.3)
ax.set_title('All peaks of H$_h \\cdots$O$_w$ wihtout noise')
ax.set_ylabel('$g_\\text{curve}(r)$')
ax.set_xlabel('r /Å');
```

## Dynamics
### MSD

```python
lags, msd, _ = traj.msd("H2O", dt=dt * dump_interval)
lags_ref, msd_ref, _ = traj_ref.msd("H2O", dt=dt * dump_interval)

mask, onset, alpha, r2 = diffusive_window(lags, msd)
print(f"Diffusive regime from {onset:.1f} fs  |  α = {alpha:.3f}  |  R² = {r2:.4f}")

mask_ref, onset_ref, alpha_ref, r2_ref = diffusive_window(lags_ref, msd_ref)
print(f"Diffusive regime from {onset_ref:.1f} fs  |  α = {alpha_ref:.3f}  |  R² = {r2_ref:.4f}")

# fit D over the detected window
coeffs = np.polyfit(lags[mask], msd[mask], 1)
D = coeffs[0] / 6 * 1e-20 / (dt * dump_interval * 1e-15)   # Å²/fs → m²/s
print(f"D = {D:.3e} m²/s")

coeffs_ref = np.polyfit(lags_ref[mask_ref], msd_ref[mask_ref], 1)
D_ref = coeffs_ref[0] / 6 * 1e-20 / (dt * dump_interval * 1e-15)   # Å²/fs → m²/s
print(f"D = {D_ref:.3e} m²/s")
```

```python

fig, ax = plt.subplots()
ax.plot(lags, msd, label='H20')
ax.plot(lags_ref, msd_ref, label='H20 reference')
ax.axvline(onset)
ax.axvline(onset_ref)
ax.set_yscale('log')
ax.set_xscale('log')
ax.legend()
ax.plot([0, 1], [0, 1], transform=ax.transAxes, color='0.5', ls='--')
#plt.xlabel("Lag (frames)"); plt.ylabel("MSD (Å²)"); plt.title("H₂O MSD")
```

H₂O in the OH• system gives D = 7.32×10⁻¹⁰ m²/s (diffusive onset ~21 ps). Pure water gives D = 7.54×10⁻¹⁰ m²/s (onset ~19 ps). The two values are within 3 % of each other.

```python
from chempiler.vanhove import van_hove

lag_vals = [2500, 5000, 7500, 10000, 12500]
r_vhw, pw, r_peaks = van_hove(traj.frames, "H2O", lags=lag_vals, return_peaks=True)
r_vhw_ref, pw_ref, r_peaks_ref = van_hove(traj_ref.frames, "H2O", lags=lag_vals, return_peaks=True)

Ds = []
for i, lag in enumerate(lag_vals):
    plt.plot(r_vhw, pw[i], color=cset[i], label=f"τ = {lag} frames")
    plt.plot(r_vhw_ref, pw_ref[i], color=cset[i], ls='--', label=f"Reference")
    #plt.yscale('log')
    #plt.xscale('log')
    #plt.xlim(0, 3)

plt.ylabel('$4 \\pi r^2 G(r,\\tau)$')
plt.legend()
```

Displacement distributions broaden with lag time in both systems. The OH• and pure water distributions are nearly identical at equivalent lags, consistent with the similar diffusion coefficients.

```python
from chempiler import van_hove_heatmap
from chempiler.vanhove import van_hove

lag_vals_hm = np.unique(np.round(np.logspace(np.log10(100), np.log10(15000), 50)).astype(int))

r_hm_w,   p_hm_w   = van_hove(traj.frames,     "H2O", lags=lag_vals_hm)
r_hm_ref, p_hm_ref = van_hove(traj_ref.frames, "H2O", lags=lag_vals_hm)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
van_hove_heatmap(r_hm_w,   p_hm_w,   lag_vals_hm, ax=axes[0])
van_hove_heatmap(r_hm_ref, p_hm_ref, lag_vals_hm, ax=axes[1])

axes[0].set_title("H₂O+OH•")
axes[1].set_title("H₂O")
for ax in axes:
    ax.set_xlabel("r /Å")
    ax.set_ylabel("lag (frames)")
plt.tight_layout()
```

The peak position shifts to larger r with increasing lag in both systems. The two heatmaps show no qualitative difference, consistent with similar H₂O diffusion in the OH• system and pure water.

## 1 — OH lifetime

`lifetime_segments` returns every contiguous interval where HO is present as a
`(start, end)` half-open tuple. The strip plot shows *when* HO exists; the
histogram shows *how long* each occurrence lasts.

```python
ho_segs = traj.lifetime_segments("HO")
seg_lengths = np.array([e - s for s, e in ho_segs])
total_ho_frames = int(seg_lengths.sum())
frac_ho = total_ho_frames / len(traj.frames)

print(f"Lifetime segments       : {len(ho_segs)}")
print(f"Total frames with HO•    : {total_ho_frames} / {len(traj.frames)} ({100*frac_ho:.1f}%)")
print(f"Segment length  min     : {seg_lengths.min()}")
print(f"               median   : {np.median(seg_lengths):.0f}")
print(f"               mean     : {seg_lengths.mean():.1f}")
print(f"               max      : {seg_lengths.max()}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 3.5))

for s, e in ho_segs:
    axes[0].axvspan(s, e, color="tomato", alpha=0.7, lw=0)
axes[0].set_xlim(0, len(traj.frames))
axes[0].set_yticks([])
axes[0].set_xlabel("Frame")
axes[0].set_title(f"HO• present in {len(ho_segs)} segments ({100*frac_ho:.1f}% of trajectory)")

max_len = int(seg_lengths.max())
bins = np.logspace(0, np.log10(max_len + 1), 30)
axes[1].hist(seg_lengths, bins=bins, color="tomato", edgecolor="white", lw=0.4)
axes[1].set_xscale("log")
axes[1].axvline(np.median(seg_lengths), color="0.3", lw=1.5, ls="--",
                label=f"median = {np.median(seg_lengths):.0f} frames")
axes[1].set_xlabel("Segment length (frames, log scale)")
axes[1].set_ylabel("Count")
axes[1].set_title("HO• lifetime distribution")
axes[1].legend()

plt.tight_layout()
```

HO• is present in 100 % of frames as a single uninterrupted segment spanning the full trajectory.

## 4 — Proton transfer kinetics

### Overall rate and time-resolved activity

A uniform hop rate across the trajectory implies uncorrelated (Poisson) transfer.
Bursts or gaps suggest cooperative events or trajectory artefacts.

```python
result     = atom_hop(traj.frames)
result_ref = atom_hop(traj_ref.frames)
```

```python
transitions   = result["transitions"]
n_hops        = result["n_transitions"]
hop_frame_arr = np.array([t[0] for t in transitions], dtype=int)
rate_overall  = n_hops / len(traj.frames)
rate_ref      = result_ref["n_transitions"] / len(traj_ref.frames)
dt_per_frame  = dt * dump_interval * 1e-3  # ps per frame

print(f"Total hops (OH• system) : {n_hops}")
print(f"Hop rate   (OH• system) : {rate_overall:.4f} /frame  ({rate_overall / dt_per_frame:.2f} ps⁻¹)")
print(f"Hop rate   (pure water) : {rate_ref:.4f} /frame  ({rate_ref / dt_per_frame:.2f} ps⁻¹)")

window = 200
step   = 50
t_cen  = np.arange(window // 2, len(traj.frames) - window // 2, step)
w_rate = np.array([
    ((hop_frame_arr >= c - window // 2) & (hop_frame_arr < c + window // 2)).sum() / window
    for c in t_cen
])

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(hop_frame_arr if n_hops > 0 else [0], bins=max(1, min(50, n_hops)),
             color="steelblue", edgecolor="white", lw=0.4)
axes[0].set_xlabel("Frame")
axes[0].set_ylabel("Events")
axes[0].set_title(f"Proton transfer events  ({n_hops} total)")

axes[1].plot(t_cen, w_rate, color="steelblue", lw=1.1)
axes[1].axhline(rate_overall, color="tomato", lw=1.2, ls="--",
                label=f"OH⁻ system  {rate_overall:.4f} /frame")
axes[1].axhline(rate_ref, color="0.5", lw=1.2, ls=":",
                label=f"pure water  {rate_ref:.4f} /frame")
axes[1].set_xlabel("Frame")
axes[1].set_ylabel("Hop rate (events / frame)")
axes[1].set_title(f"Windowed hop rate  (window = {window} frames)")
axes[1].legend()
plt.tight_layout()
```

No proton hops are detected in the OH• system. The hop rate is zero.

### Hop rate conditioned on HO presence

```python
ho_present = np.zeros(len(traj.frames), dtype=bool)
for s, e in ho_segs:
    ho_present[s:e] = True

frames_with_ho    = int(ho_present.sum())
frames_without_ho = len(traj.frames) - frames_with_ho

hops_in_ho  = int(ho_present[hop_frame_arr].sum()) if n_hops > 0 else 0
hops_out_ho = n_hops - hops_in_ho

rate_in  = hops_in_ho  / frames_with_ho    if frames_with_ho    > 0 else 0.0
rate_out = hops_out_ho / frames_without_ho if frames_without_ho > 0 else 0.0

print(f"Frames with HO present  : {frames_with_ho:>5}  ({100*frames_with_ho/len(traj.frames):.1f}%)")
print(f"Frames without HO       : {frames_without_ho:>5}")
print()
print(f"Hops when HO present    : {hops_in_ho:>3} of {n_hops}  →  rate = {rate_in:.4f} /frame  ({rate_in / dt_per_frame:.3f} ps⁻¹)")
print(f"Hops when HO absent     : {hops_out_ho:>3} of {n_hops}  →  rate = {rate_out:.4f} /frame  ({rate_out / dt_per_frame:.3f} ps⁻¹)")
if rate_out > 0 and rate_in > 0:
    print(f"\nRate ratio (absent / present): {rate_out / rate_in:.1f}×")
```

HO• is present in all 20001 frames. No proton hops are detected, so the conditioned rates are both zero and no rate ratio is defined.

## 5 — Hop-to-OH distance

`hop_species_distances` records, for each hop event where HO is present in that
frame, the minimum-image distance from the hopping hydrogen to the nearest HO
molecule centre of mass. This tests whether proton transfers happen close to or
far from the hydroxide.

```python
d = traj.hop_species_distances(result, formula="HO", reference="H")

print(f"Total hop events          : {d['n_hops_total']}")
print(f"Events with HO present    : {d['n_measured']}")
if d['n_measured'] > 0:
    print(f"Mean H → HO distance      : {d['mean']:.2f} Å")
    print(f"Min / Max                 : {d['distances'].min():.2f} / {d['distances'].max():.2f} Å")

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.hist(d["distances"], bins=20, color="steelblue", edgecolor="white", lw=0.4)
    ax.axvline(d["mean"], color="tomato", lw=1.5, ls="--",
               label=f"mean = {d['mean']:.2f} Å")
    ax.set_xlabel("Distance from hopping H to nearest HO (Å)")
    ax.set_ylabel("Events")
    ax.set_title("Proton hop → HO distance")
    ax.legend()
    plt.tight_layout()
    plt.show()
else:
    print("No hop events coincided with HO being present.")
```

No hop events are detected, so no distance distribution is available.

## 6 — Solvation shell transitions (ligand exchange)

`ligand_exchange` tracks changes in molecular formula for every molecule across
every frame. Filtering for events that involve HO on either side shows how HO
forms, evolves, and eventually returns to H₂O or picks up additional atoms to
form larger clusters.

```python
lex    = ligand_exchange(traj.frames)
ho_lex = [(fi, mol, f, t)
          for fi, mol, f, t in lex["transitions"]
          if "HO" in (f, t)]
ho_types = Counter((f, t) for _, _, f, t in ho_lex)

print(f"Total ligand-exchange events    : {lex['n_transitions']}")
print(f"Events involving HO             : {len(ho_lex)}")
print()
print("Transition types:")
for (a, b), count in sorted(ho_types.items(), key=lambda x: -x[1]):
    direction = "← HO gains H" if b == "HO" else "→ HO loses H"
    print(f"  {a:<8} → {b:<8} : {count:>4}   {direction}")
```

```python
if ho_types:
    items  = sorted(ho_types.items(), key=lambda x: -x[1])
    labels = [f"{a} → {b}" for (a, b), _ in items]
    counts = [c for _, c in items]

    colors = ["steelblue" if a == "HO" else "tomato" for (a, b), _ in items]

    fig, ax = plt.subplots(figsize=(9, max(3, 0.45 * len(items))))
    ax.barh(range(len(labels)), counts, color=colors, edgecolor="white")
    ax.set_yticks(range(len(labels)))
    ax.set_yticklabels(labels, fontsize=9)
    ax.invert_yaxis()
    ax.set_xlabel("Events")
    ax.set_title("Molecular transitions involving HO\nBlue = HO source  |  Red = HO product")

    from matplotlib.patches import Patch
    ax.legend(handles=[
        Patch(color="tomato",    label="HO formed  (X → HO)"),
        Patch(color="steelblue", label="HO consumed (HO → X)"),
    ], loc="lower right")
    plt.tight_layout()
```

No ligand-exchange events involving HO• are detected across the full trajectory.
